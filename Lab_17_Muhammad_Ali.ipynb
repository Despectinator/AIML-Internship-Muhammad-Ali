## Task 1 — Vague-to-Specific Rewrite

### Vague Prompt

```
Write about productivity.
```

**Model Response (Vague Prompt):**

Productivity is the measure of how efficiently a person or organization completes tasks and achieves goals. It can be improved through better time management, prioritization, reducing distractions, and taking regular breaks. Many people also find that setting clear goals and using tools like to-do lists or calendars helps them stay focused. Ultimately, productivity is about working smarter, not just harder, and finding a rhythm that allows for consistent progress without burnout.

### Specific Prompt
Rewritten using instruction, context, input/topic, format, and constraints:

```
Write a 150-word article explaining three practical time-management
techniques for university students preparing for final exams. Use an
encouraging, informal tone and avoid generic advice like "just make a
to-do list." End with one specific action the reader can take today.
```

**Model Response (Specific Prompt):**

Exam season can feel overwhelming, but a few targeted habits make a real difference. First, try "time-blocking": instead of a vague plan to "study biology," schedule a fixed 90-minute block for it, treating it like an appointment you can't skip. Second, use the Pomodoro technique with a twist — study in 25-minute sprints, but write down the one thing that distracted you each time you pause, so you learn your own patterns instead of just fighting them blindly. Third, batch your low-effort tasks (printing notes, replying to group-chat messages) into one 15-minute slot instead of letting them interrupt deep study whenever they pop up. None of this requires new apps or willpower you don't already have — it just protects the focus you do have. Today: pick your hardest subject and block out one 90-minute session for it before you do anything else.

### Comparison

The vague prompt produced a generic, encyclopedia-style overview of productivity with no clear audience or actionable takeaway — the kind of answer that could apply to anyone and therefore helps no one in particular. The specific prompt, by contrast, produced content tailored to a defined audience (university students), a defined context (exam season), a defined length, a defined tone, and an explicit instruction to end with a concrete action — and the output honored every one of those constraints. This illustrates the core lesson of prompt engineering: the model isn't smarter or dumber between the two prompts, it simply has less to guess when the instruction, context, format, and constraints are spelled out.

Few-shot prompt with three labeled examples, used to classify customer support messages into Billing, Technical, or General Inquiry:

```
Classify the following customer support message as Billing, Technical,
or General Inquiry.

Message: "Absolutely love this product, works perfectly!"
Category: General Inquiry

Message: "I was charged twice for my subscription this month."
Category: Billing

Message: "The app crashes every time I try to upload a photo."
Category: Technical

Message: "{new message}"
Category:
```

### Test Messages and Results

1. "My invoice shows a different amount than what I agreed to at signup." → **Category: Billing**
2. "I can't log in — the page just spins and then times out." → **Category: Technical**
3. "Do you have a mobile app available for Android?" → **Category: General Inquiry**

All three unseen messages were classified correctly. ## Task 3 — Role Prompting Comparison

### No Role Assigned

```
Should I invest in index funds or individual stocks?
```

**Model Response (No Role):**

Both options have trade-offs. Index funds offer broad diversification, low fees, and historically consistent long-term returns, making them a popular choice for investors who don't want to spend time researching individual companies. Individual stocks can offer higher potential returns and let you invest directly in companies you believe in, but they carry higher risk, require more research and monitoring, and are harder to diversify with a small amount of money. Many investors use a mix of both: a core of index funds for stability, with a smaller portion in individual stocks for those who enjoy hands-on investing. The right choice depends on your risk tolerance, time horizon, and how much time you want to spend managing your portfolio.

### Role Assigned

```
You are a conservative financial advisor for retirees. Should I invest
in index funds or individual stocks?
```

**Model Response (With Role):**

For someone in or near retirement, I would generally steer away from individual stock-picking and lean strongly toward index funds — particularly low-cost, broad-market or dividend-focused funds, paired with a solid allocation to bonds. At this stage, capital preservation and predictable income matter far more than chasing higher returns. Individual stocks expose you to company-specific risk: if one holding drops sharply, you may not have the years remaining to recover before needing that money for living expenses. A diversified index fund spreads that risk across hundreds or thousands of companies. I'd recommend reviewing your withdrawal timeline with an advisor and considering a bond ladder or annuity for guaranteed income, with index funds providing modest, steady growth to help your savings keep pace with inflation.

### Comparison

The no-role response is balanced and neutral, presenting both options evenhandedly and leaving the decision to the reader. The role-assigned response is noticeably more directive and risk-averse: it takes a clear position (favor index funds, avoid individual stocks), and grounds its reasoning in concerns specific to retirees — capital preservation, income needs, and limited time to recover from losses — rather than general investing pros and cons. The vocabulary also shifts to reflect the persona ("capital preservation," "withdrawal timeline," "bond ladder") in a way the neutral response never does. This shows how a role/persona doesn't just change tone — it changes which considerations the model foregrounds and how confidently it commits to a recommendation.

## Task 4 — Chain-of-Thought Test

Problem used for both prompts:

```
A train leaves Station A traveling at 60 mph. Two hours later, a
second train leaves Station A on the same track, traveling in the
same direction at 90 mph. How many hours after the FIRST train left
does the second train catch up to it?
```

### Direct Answer Only

```
[Problem above] Give only the final answer, no explanation.
```

**Model Response (Direct Answer):**

4 hours.

This answer is incorrect — it looks like the 90 mph vs. 60 mph gap was used without accounting for the first train's two-hour head start, so the model appears to have jumped to a plausible-looking number rather than actually working through the relative positions of both trains.

### Step-by-Step (Chain-of-Thought)

```
[Problem above] Think through this step by step before giving your final answer.
```

**Model Response (Chain-of-Thought):**

**Step 1:** In the 2 hours before the second train leaves, the first train travels 60 mph × 2 h = 120 miles. So when the second train starts, it is 120 miles behind.

**Step 2:** Once both trains are moving, the second train gains on the first at a rate of (90 − 60) = 30 mph relative speed.

**Step 3:** Time for the second train to close a 120-mile gap at 30 mph = 120 ÷ 30 = 4 hours after the second train departs.

**Step 4:** The question asks how long after the FIRST train left, so add the 2-hour head start: 4 + 2 = 6 hours.

**Final answer:** 6 hours after the first train left.

### Comparison

The direct-answer prompt produced a confident but wrong result (4 hours), most likely because it answered based on the relative-speed step alone and skipped accounting for the head start or what "after the first train left" actually asks for. The chain-of-thought prompt produced the correct result (6 hours) by explicitly breaking the problem into the head-start distance, the closing speed, and the catch-up time, then correctly converting back to the question's reference point. This is a clear, concrete example of the point made in Lab 1 and this lab: LLMs don't guarantee correct reasoning by default, but explicitly asking for step-by-step reasoning gives the model room to work through intermediate quantities instead of pattern-matching straight to a superficially plausible number.

## Task 5 — Iterative Refinement Log

Real task: drafting an email to my manager requesting a short deadline extension on a project.

### Version 1 — Initial Prompt

```
Write an email asking for more time on a project.
```

**Model Response (V1):**

```
Subject: Extension Request

Hi,

I hope this message finds you well. I am writing to request additional time to complete my current project. I have encountered some challenges that have affected my timeline, and I believe a short extension would allow me to deliver higher-quality work.

Please let me know if this is possible.

Thank you for your understanding.

Best regards
```

**What was missing:** no specific project name, no reason for the delay, no proposed new deadline, and the tone is generic enough to have been written about literally any project — a manager reading this would likely just reply asking for these exact details.

### Version 2 — First Refinement

```
Write a professional email to my manager asking for a 3-day extension
on the Q3 report deadline. Mention that the delay is due to waiting on
data from another team, and propose a new deadline of Friday instead
of Tuesday.
```

**Model Response (V2):**

```
Subject: Request for Extension on Q3 Report

Hi [Manager's Name],

I wanted to reach out regarding the Q3 report, which is currently due on Tuesday. I'm still waiting on a key dataset from the finance team, and without it I won't be able to complete the analysis to the standard I'd like by the original deadline.

I'd like to request a short 3-day extension, moving the deadline to Friday. This would give me enough buffer to receive the data, finish the analysis, and do a proper review before submitting.

Please let me know if Friday works on your end, or if there's a different timeline you'd prefer. I'm happy to share a progress update in the meantime if that's helpful.

Thank you for understanding,
[Your Name]
```

**What changed and why:** added the specific project (Q3 report), the reason (waiting on data from finance), and the proposed new date (Friday instead of Tuesday) — directly addressing the vagueness gap from V1. Remaining gap: it's now well-specified but quite long for what should be a quick, low-friction request, and the tone reads slightly more formal/stiff than needed for an internal manager email.

### Version 3 — Second Refinement

```
Write a concise, polite email (under 120 words) with a subject line, to
my manager, asking for a 3-day extension on the Q3 report deadline
(from Tuesday to Friday) because I'm still waiting on data from the
finance team. Keep the tone professional but not overly formal.
```

**Model Response (V3 — Final):**

```
Subject: Q3 Report — Requesting a Short Extension

Hi [Manager's Name],

Quick heads-up: I'm still waiting on data from the finance team for the Q3 report, so I'd like to request a 3-day extension — moving the deadline from Tuesday to Friday. That should give me enough time to finish the analysis properly once the data comes in.

Let me know if Friday works, or if you'd prefer a different timeline.

Thanks,
[Your Name]
```

**What changed and why:** added an explicit length constraint (under 120 words) and a tone constraint (professional but not overly formal) to fix the two gaps identified after V2. The result is noticeably tighter and reads like something a manager could scan and approve in seconds, while still including every essential detail (which report, how long, why, and the proposed new date).

### Summary of the Refinement Process

Each iteration targeted one specific, observed gap rather than starting over: V1 → V2 added missing specifics (what, why, when); V2 → V3 added a length and tone constraint to fix over-formality and length. This mirrors the lab's core point about iterative refinement — diagnose exactly what's wrong with an output, then adjust only the part of the prompt responsible for that gap.
